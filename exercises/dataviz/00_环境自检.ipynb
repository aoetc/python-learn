```python
import sys
import matplotlib
import pandas as pd
print('Python:', sys.version.split()[0])
print('可执行文件:', sys.executable)
print('matplotlib:', matplotlib.__version__)
print('pandas:', pd.__version__)
print('字体优先:', matplotlib.rcParams['font.sans-serif'][:3])
```

```python
import matplotlib.pyplot as plt

months = ['1月', '2月', '3月', '4月']
spend = [800, 1000, 500, 900]

plt.figure(figsize=(6, 3.5))
plt.bar(months, spend, color='#3B82F6')
plt.title('宿舍月花销（自检图）')
plt.ylabel('元')
plt.tight_layout()
plt.show()
```
